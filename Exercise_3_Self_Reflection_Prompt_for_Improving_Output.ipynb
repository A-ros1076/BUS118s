# Exercise 3: Self-Reflection Prompt for Improving Output

**Goal:** Ask Gemini to generate a product description, critique its own output against explicit requirements, then produce an improved version.

| Step | Task | Input | Output |
|------|------|-------|--------|
| 1 | Generate initial product description | Product specs | First draft |
| 2 | Self-critique | Step 1 output + criteria | Scored critique per criterion |
| 3 | Improve | Step 1 output + Step 2 critique | Revised, improved description |

**Tool:** Google Colab + Gemini API (`gemini-2.5-flash`)

```python
!pip install google-generativeai -q
```

```python
import google.generativeai as genai
from google.colab import userdata

GOOGLE_API_KEY = userdata.get('GEMINI_API_KEY')
genai.configure(api_key=GOOGLE_API_KEY)
model = genai.GenerativeModel('gemini-2.5-flash')

def ask_gemini(prompt: str) -> str:
    """
    Sends a prompt to Gemini and returns the text response.

    Args:
        prompt: The full prompt string to send.

    Returns:
        Gemini's response as a stripped string.
    """
    try:
        response = model.generate_content(prompt)
        return response.text.strip()
    except Exception as e:
        print(f"An error occurred: {e}")
        return "Request failed."

print("Gemini client ready.")
```

## Self-Reflection Loop

All three steps run in one cell. The initial description from Step 1 feeds into the critique in Step 2. Both feed into the improved version in Step 3.

```python
# Product specs for Gemini to write a description from
PRODUCT_SPECS = """
Product: NovaBud X1 Wireless Earbuds
- Active noise cancellation (ANC)
- 30-hour total battery life (8hr buds + 22hr case)
- Bluetooth 5.3
- Touch controls
- Built-in microphone for calls
- Compatible with iOS and Android
- Available in: Midnight Black, Pearl White, Navy Blue
- Price: $79.99
"""

# Critique criteria used in Step 2
CRITERIA = """
1. Clarity: Is it easy to read with no unnecessary jargon?
2. Length: Is it 60 words or fewer?
3. Compelling hook: Does the opening line grab attention or speak to a customer pain point?
4. Key features highlighted: Does it mention ANC, battery life, and connectivity?
5. Customer benefit focus: Are features framed around what the customer gains, not just listed as specs?
6. Call to action: Does it end with a line encouraging the customer to act?
"""

# ============================================================
# STEP 1 — Generate initial description (minimal prompt = flawed first draft)
# ============================================================
print("=" * 60)
print("STEP 1: INITIAL PRODUCT DESCRIPTION")
print("=" * 60)

step1_prompt = f"""
Write a short product description for the following tech gadget.

Product specs:
{PRODUCT_SPECS}
"""

step1_output = ask_gemini(step1_prompt)
print(step1_output)

# ============================================================
# STEP 2 — Self-critique against explicit criteria
# ============================================================
print("\n" + "=" * 60)
print("STEP 2: SELF-CRITIQUE")
print("=" * 60)

step2_prompt = f"""
You are a copywriting critic. Review the product description below against each criterion.
For each criterion, respond with PASS or FAIL and a one-sentence explanation.

Product description to review:
\"{step1_output}\"

Criteria:
{CRITERIA}

Format your response as a numbered list matching the criteria above.
Be critical and specific — do not give a PASS unless it is clearly earned.
"""

step2_output = ask_gemini(step2_prompt)
print(step2_output)

# ============================================================
# STEP 3 — Generate improved description using the critique
# ============================================================
print("\n" + "=" * 60)
print("STEP 3: IMPROVED PRODUCT DESCRIPTION")
print("=" * 60)

step3_prompt = f"""
You are a copywriting editor. Using the critique below, rewrite the product description
to fix every FAIL and maintain every PASS.

Original description:
\"{step1_output}\"

Critique:
{step2_output}

Requirements for the improved version:
{CRITERIA}

Return only the improved description — no explanation, no labels.
"""

step3_output = ask_gemini(step3_prompt)
print(step3_output)

# ============================================================
# COMPARISON — Before and after side by side
# ============================================================
print("\n" + "=" * 60)
print("COMPARISON")
print("=" * 60)
before_wc = len(step1_output.split())
after_wc = len(step3_output.split())
print(f"BEFORE ({before_wc} words): {step1_output}")
print(f"\nAFTER ({after_wc} words): {step3_output}")
```
